# 02 · Limpieza de datos

Se aplican las decisiones detectadas en el notebook 01. La lógica vive en `src/data.py` para poder reutilizarla y probarla.

```python
# configuracion 
import sys
from pathlib import Path

import pandas as pd

sys.path.append(str(Path.cwd().parent))
from src.data import cargar_raw, limpiar_datos, guardar_procesado

df_raw = cargar_raw()
df_limpio = limpiar_datos(df_raw)
print(df_raw.shape, "->", df_limpio.shape)
```

```python
# comprobar los cambios

print("Tipo de TotalCharges antes:", df_raw["TotalCharges"].dtype)
print("Tipo de TotalCharges después:", df_limpio["TotalCharges"].dtype)
print("Nulos después:", df_limpio.isna().sum().sum())
print("Valores con comillas:", sum(df_limpio[c].astype(str).str.contains("'").sum()
      for c in df_limpio.columns))
print(df_limpio["Contract"].unique())
print(df_limpio["Churn"].value_counts())
```

```python
# duplicados

print("Filas duplicadas:", df_limpio.duplicated().sum())
```

```python
# Guardar el dataset limpio

ruta = guardar_procesado(df_limpio)
print("Guardado en", ruta)

```

## Decisiones de limpieza
- `TotalCharges` pasó a numérico. Los 11 vacíos son clientes nuevos (`tenure = 0`), se imputaron con 0 en lugar de eliminarlos.
- Se quitaron las comillas y espacios sobrantes de las columnas de texto.
- `Churn` se convirtió a binario (1 = abandonó).
